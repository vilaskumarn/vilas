```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('encodeing_life_expectancy.csv')
```

```python
df
```

```python
df.rename(columns={'Life expectancy ':'Life expectancy'},inplace=True)
```

```python
Y = df['Life expectancy']
X = df.drop(columns='Life expectancy',axis=1)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.20,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
Scaler = StandardScaler()
```

```python
X_train_sclar = Scaler.fit_transform(X_train)
X_test_sclar = Scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
Liner_model = LinearRegression()
KNN_model = KNeighborsRegressor(n_neighbors=5)
DecisionTree_model = DecisionTreeRegressor(random_state=42)
random_model = RandomForestRegressor(random_state=42)
```

```python
# Cross valuditon
```

```python
from sklearn.model_selection import cross_val_score
```

```python
model_dir ={}
models = [Liner_model,KNN_model,DecisionTree_model,random_model]
for model in models:
    score = cross_val_score(model,X_train_sclar,Y_train,scoring='neg_mean_squared_error',cv=5)
    print(f"{model}:{abs(score)} mean_value::{np.mean(abs(score))}")
    model_dir[model]=float(np.mean(abs(score)))
value =(model_dir.values())
best_model = min(model_dir, key=model_dir.get)
print(f"the best model::{best_model}")
print(f"the less error model::{min(value)}")

    
```
