```python
Q1 = df['AMT_CREDIT'].quantile(0.25)
Q3 = df['AMT_CREDIT'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['AMT_CREDIT'] < lower_bound) | (df['AMT_CREDIT'] > upper_bound)]

print(len(outliers))

# print(f"Lower Threshold = {lower_bound}")
# print(f"Upper Threshold = {upper_bound}")

```

```python
start_memory = df.memory_usage(deep=True).sum() / 1024**2
```

```python
print(f"{round(start_memory,3)} Mb")
```

```python
c_min = df['AMT_CREDIT'].min()
c_max = df['AMT_CREDIT'].max()

if c_min >= np.iinfo(np.int8).min and c_max <= np.iinfo(np.int8).max:
    df['AMT_CREDIT'] = df['AMT_CREDIT'].astype(np.int8)
elif c_min >= np.iinfo(np.int16).min and c_max <= np.iinfo(np.int16).max:
    df['AMT_CREDIT'] = df['AMT_CREDIT'].astype(np.int16)
elif c_min >= np.iinfo(np.int32).min and c_max <= np.iinfo(np.int32).max:
    df['AMT_CREDIT'] = df['AMT_CREDIT'].astype(np.int32)
else:
    df['AMT_CREDIT'] = df['AMT_CREDIT'].astype(np.int64)
```

```python
end_memory = df.memory_usage(deep=True).sum() / 1024**2
```

```python
print(f"{round(end_memory,3)} Mb")
```

```python
per=((start_memory-end_memory)/start_memory)*100
```

```python
print(f"{round(per,3)}%")
```
