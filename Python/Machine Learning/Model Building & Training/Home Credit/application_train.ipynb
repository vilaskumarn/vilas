## Load Data

```python
import pandas as pd

df = pd.read_csv("C:/Users/vilas/Documents/0-100/AI,ML & DL/home-credit-default-risk/application_train.csv")

df.shape
```

## Preprocessing

```python
## Null Values

df.isna().sum().sum()
```

```python
# Filling null values of int and float dtypes

for i in df.select_dtypes(include=['int','float']).columns:
    df.fillna(df[i].median(), inplace=True)
    
# Filling null values of char dtypes

for i in df.select_dtypes(include='object').columns:
    df.fillna(df[i].value_counts().index[0], inplace=True)
```

```python
df.isna().sum().sum()
```

```python
# Label Encoding

df.select_dtypes(include='object').columns
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for i in df.select_dtypes(include='object').columns:
    df[i] = df[i].astype(str)
    le.fit(df[i])
    transformed = le.transform(df[i])
    df[i] = transformed
```

```python
df.head()
```

## Model Building

```python
df['TARGET'].value_counts()

# We have imbalanced data
```

```python
# First split to train and test data

X = df.drop('TARGET', axis=1)

y = df['TARGET']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

dt = DecisionTreeClassifier()
rf = RandomForestClassifier()

dt.fit(X_train,Y_train)

y_pred1 = dt.predict(X_test)

rf.fit(X_train,Y_train)

y_pred2 = rf.predict(X_test)
```

## Model Validation

```python
from sklearn.metrics import accuracy_score, confusion_matrix

print(f"Confusion Matrix of DecisionTree Model : \n{confusion_matrix(Y_test,y_pred1)}")
print(f"Confusion Matrix of DecisionTree Model : {accuracy_score(Y_test,y_pred1):.2f}")
print(f"Accuracy of RandomForest Model :\n {confusion_matrix(Y_test,y_pred2)}")
print(f"Accuracy of RandomForest Model : {accuracy_score(Y_test,y_pred2):.2f}")
```

```python
from sklearn.metrics import classification_report

print(f"Classification Report of DecisionTree : \n{classification_report(Y_test,y_pred1)}")
print('==========================================================')
print(f"Classification Report of Random Forest : \n{classification_report(Y_test,y_pred2)}")
```
