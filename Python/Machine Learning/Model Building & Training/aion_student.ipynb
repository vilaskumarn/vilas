```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str)

    le.fit(df[col])

    trans = le.transform(df[col])
    df[col] = trans
```

```python
df.head()
```

```python
df['Paid_Subscription'] = le.fit_transform(df['Paid_Subscription'])
```

```python
df.head()
```

```python
x = df.drop('Burnout_Risk_Level', axis=1)
y = df['Burnout_Risk_Level']

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)

```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

dt = DecisionTreeClassifier()
rf = RandomForestClassifier()

l = [dt,rf]
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score

for i in l:
    i.fit(x_train,y_train)
    y_pred = i.predict(x_test)

    print(f"Confusion Matrix of {i.__class__.__name__} :\n{confusion_matrix(y_test,y_pred)}")
    print(f"Accuracy Score of {i.__class__.__name__} :\n{accuracy_score(y_test,y_pred):.2f}")
    print()
```
