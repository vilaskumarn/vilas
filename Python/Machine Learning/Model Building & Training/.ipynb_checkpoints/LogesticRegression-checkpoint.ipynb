# Setting up Project

Import Required Libraries 👇🏼

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn import preprocessing
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('bank.csv',header=0,sep=';')
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.shape
```

```python
# Displaying all the fields/columns

print(list(df.columns))
```

```python
# Eliminating Unwanted Fields
#drop columns which are not needed.

df.drop(df.columns[[0, 3, 5, 8, 9, 10, 11, 12, 13, 14]], axis = 1, inplace = True)
```

```python
df.head()
```

```python
# Encoding the data

data = pd.get_dummies(df,columns=['job','marital','default','housing','loan','poutcome'])
```

```python
data.head()
```

```python
data.columns
```

```python
data
```

```python
data.drop(data.columns[[12, 16, 18, 21, 24]], axis=1, inplace=True)
```

```python
data.columns
```

```python
# Splitting Data

X = data.iloc[:,1:]
X.head()
```

```python
y = data.iloc[:,0]
y.head()
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,y,random_state=0)
```

```python
classifier = LogisticRegression(solver='lbfgs',random_state=0)
```

```python
classifier.fit(X_train,Y_train)
```

```python
predicted_y = classifier.predict(X_test)
```

```python
predicted_y
```

```python
for x in range(len(predicted_y)):
   if (predicted_y[x] == 'yes'):
      print(x, end="\t")
```

```python
classifier.score(X_test,Y_test)
```
